```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import GradientBoostingClassifier
import numpy as np
```

# Uncertainty Estimates

Uncertainty estimates express how confident a classifier is in its prediction. In scikit-learn, uncertainty can be obtained using decision_function, which returns real-valued confidence scores with arbitrary scale, or predict_proba, which returns normalized probabilities for each class. These estimates are crucial in applications where different types of classification errors have different real-world consequences.

```python
iris = load_iris()
X = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
gbrt = GradientBoostingClassifier(learning_rate=0.01, random_state=0)
gbrt.fit(X_train, y_train)
```

# Decision Function

```python
print("Decision function shape:", gbrt.decision_function(X_test).shape)
# First few entries of the decision function
print("Decision function:")
print(gbrt.decision_function(X_test)[:6, :])
```

In the multiclass case, the decision_function has the shape (n_samples, n_classes) and each column provides a “certainty score” for each class, where a large score means that a class is more likely and a small score means the class is less likely.

We can recover the predictions from these scores by finding the maximum entry for each data point.

```python
print("Argmax of decision function:")
print(np.argmax(gbrt.decision_function(X_test), axis=1)) #returns the index of the largest value row by row
print("Predictions:")
print(gbrt.predict(X_test))
```

# Predicting Probabilities

The output of predict_proba has the same shape, (n_samples, n_classes). The probabilities for the possible classes for each data point sum to 1.

```python
# First few entries of predict_proba
print("Predicted probabilities:")
print(gbrt.predict_proba(X_test)[:6])
# show that sums across rows are one
print("Sums:", gbrt.predict_proba(X_test)[:6].sum(axis=1))
```

We can again recover the predictions by computing the argmax of predict_proba.

```python
print("Argmax of predicted probabilities:")
print(np.argmax(gbrt.predict_proba(X_test), axis=1))
print("Predictions:")
print(gbrt.predict(X_test))
```
